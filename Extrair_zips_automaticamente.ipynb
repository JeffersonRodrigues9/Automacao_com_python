```python
# Automatização do processo de descompactação de arquivos ZIP.
# O script percorre automaticamente uma pasta principal e suas subpastas,
# extraindo todos os arquivos ZIP encontrados e removendo os arquivos compactados após a extração.
# Também identifica e descompacta ZIPs contidos dentro de outros ZIPs até finalizar todo o processo.

import os
import zipfile
from tqdm import tqdm

def extrair_zip(caminho_zip, pasta_destino):
    try:
        with zipfile.ZipFile(caminho_zip, 'r') as zip_ref:
            zip_ref.extractall(pasta_destino)
        return True
    except Exception as e:
        print(f"Erro ao extrair {caminho_zip}: {e}")
        return False

def encontrar_zips(pasta_principal):
    arquivos_zip = []

    for raiz, dirs, arquivos in os.walk(pasta_principal):
        for arquivo in arquivos:
            if arquivo.lower().endswith(".zip"):
                arquivos_zip.append(os.path.join(raiz, arquivo))

    return arquivos_zip

def descompactar_todos_os_zips(pasta_principal):
    while True:
        arquivos_zip = encontrar_zips(pasta_principal)

        if not arquivos_zip:
            print("\nNenhum arquivo ZIP restante encontrado.")
            break

        print(f"\n{len(arquivos_zip)} arquivo(s) ZIP encontrado(s).")

        for caminho_zip in tqdm(arquivos_zip, desc="Descompactando ZIPs"):
            pasta_destino = os.path.splitext(caminho_zip)[0]

            os.makedirs(pasta_destino, exist_ok=True)

            sucesso = extrair_zip(caminho_zip, pasta_destino)

            if sucesso:
                try:
                    os.remove(caminho_zip)
                except Exception as e:
                    print(f"Erro ao remover {caminho_zip}: {e}")

if __name__ == "__main__":
    pasta_principal = r""

    descompactar_todos_os_zips(pasta_principal)

    print("\nProcesso finalizado.")
```
